# Guide d'optimisation

## CPU

Le pourcentage d'utilisation CPU est le pourcentage de temps passé par le(s) CPU(s) à effectuer des calculs. L'attente pour les entrées/sorties n'est pas prise en compte dans le calcul de ce pourcentage, c'est la raison pour laquelle il peut être inférieur à 100%.

Pour optimiser ce pourcentage (le rapprocher de 100%), plusieurs possibilités:
- CPU < 100%: Si vous avez demandé plus d'un thread CPU, réduisez le nombre de threads d'autant. Exemple, si CPU=50% et threads=8, passez à threads=4. La tâche n'ira pas nécessairement moins vite, en revanche, demander moins de coeurs en libérera autant pour d'autres tâches.
- CPU proche de 100%: Vous n'avez pas à modifier le nombre de coeurs demandés à SLURM.
- CPU > 100%: Votre tâche a utilisé plus de coeurs que ce qui avait été déclaré. Essayez d'évaluer le nombre de coeurs qui a été utilisé et ajustez les ressources demandées en conséquence.

## Mémoire

Le pourcentage d'utilisation de la mémoire correspond au rapport entre la consommation de RAM réelle (au maximum) et la quantité de RAM requise à la soumission du job SLURM.

Pour optimiser ce pourcentage (le rapprocher de 100%), plusieurs possibilités:
- Utilisation mémoire < 100%: diminuer la quantité de mémoire requise, **à une exception près**: si la durée maximale du job est inférieure à 30 secondes, l'utilisation mémoire n'est pas représentative de la tâche. Ceci est dû à la fréquence d'échantillonage de la consommation de RAM par SLURM, qui est de 30 secondes (non modifiable).
- Utilisation mémoire > 100%: augmenter la quantité de mémoire requise. Autrement, le risque d'OOM Kill augmente significativement.
- Utilisation mémoire autour de 100%: Vous n'avez rien à optimiser.

## Durée d'exécution

Limiter la durée d'exécution d'un job permet au scheduler (SLURM en l'occurrence) de placer des jobs courts lorsqu'une fenêtre de ressources est disponible pour un temps supérieur ou égal à la durée d'exécution spécifiée. C'est le principe du backfilling.
